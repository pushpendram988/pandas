Merge

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],"B":[11,12,13,14]})
var2 = pd.DataFrame({"A":[1,2,3,5],"c":[21,22,23,24]})

m = pd.merge(var1,var2,on="A")
m
```

```python
pd.merge(var1,var2,how="inner")
```

```python
pd.merge(var1,var2,how="right")
```

```python
pd.merge(var1,var2,how="outer")
```

```python
import pandas as pd
var3 =pd.DataFrame({"A":[1,2,3,4],"B":[11,12,13,14]})
var4=pd.DataFrame({"A":[1,2,3,5],"B":[21,22,23,24]})
pd.merge(var3,var4,left_index=True,right_index=True)
```

```python
pd.merge(var3,var4,left_index=True,right_index=True,suffixes=("name","id"))
```

Key Parameter

```python

d11 = pd.DataFrame({"A":[1,2,3,4],"B":[11,12,13,14]})
d22 = pd.DataFrame({"A":[1,2,3,5],"c":[21,22,23,24]})
pd.concat([d11,d22],keys=["d11","d22"])
```

```python
pd.concat([d11,d22],axis=1,keys=["d11","d22"])
```
